### **1.6 Business Metric**

### **4.3 Duplikat, nilai null, dan `unknown`**

```python
# Memeriksa baris duplikat dan nilai kosong
print('Baris duplikat penuh :', df_raw.duplicated().sum())
print('Sel kosong (null)    :', df_raw.isna().sum().sum())
```

Tidak ada baris duplikat dan tidak ada sel kosong, jadi tidak ada baris yang perlu dihapus atau diisi karena dua hal ini.

```python
# Mengecek nilai unknown
cek = pd.DataFrame({
    'unknown': (df_raw == 'unknown').sum()
})
cek['unknown_%'] = cek['unknown'] / len(df_raw) * 100
cek[cek['unknown'] > 0]
```

Terdapat beberapa kolom yang menggunakan 'unknown' untuk menandai informasi yang tidak tersedia atau tidak diketahui.

```python
# Memerika nasabah dengan riwayat kampanye unknown
poutcome_unknown = df_raw[df_raw['poutcome'] == 'unknown']

print('Jumlah nasabah dengan poutcome = unknown :', len(poutcome_unknown))
print('  belum pernah dihubungi (pdays = -1)    :', (poutcome_unknown['pdays'] == -1).sum())
print('  pernah dihubungi (pdays bukan -1)      :', (poutcome_unknown['pdays'] != -1).sum())
```

### **4.5 Outlier**

Outlier ditandai dengan aturan IQR: nilai di bawah Q1 − 1,5 × IQR atau di atas Q3 + 1,5 × IQR.

```python
# Mendeteksi outlier
baris = []
for c in kolom_num:
    q1, q3 = df_raw[c].quantile([.25, .75])
    iqr = q3 - q1
    bawah, atas = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    luar = (df_raw[c] < bawah) | (df_raw[c] > atas)
    baris.append({'kolom': c, 'Q1': q1, 'Q3': q3, 'batas_atas': atas,
                  'jumlah_outlier': luar.sum(), 'outlier_%': luar.mean() * 100})
pd.DataFrame(baris).set_index('kolom')
```

### **4.7 Merekonstruksi Tahun**

Dataset tidak memiliki kolom tahun secara langsung, padahal periode kampanye mencakup tiga tahun. Untuk memungkinkan analisis berdasarkan waktu, tahun direkonstruksi dari urutan baris dan perubahan bulan dalam data.

Ketika urutan bulan kembali ke bulan yang lebih awal, perubahan tersebut digunakan sebagai penanda pergantian tahun. Hasil rekonstruksi kemudian diperiksa dengan membandingkan periode awal dan akhir data terhadap periode publikasi dataset.

```python
# Mengubah nama bulan menjadi angka agar urutan waktu bisa diperiksa
nomor_bulan = {'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6,
               'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12}

cek = df_raw.copy()                                
cek['bulan_num'] = cek['month'].map(nomor_bulan)   
```

```python
# Mengurutkan tahun berdasarkan urutan bulan dalam dataset
daftar_tahun = []
tahun_sekarang = 2008                              
bulan_sebelumnya = cek['bulan_num'].iloc[0]        

for bulan in cek['bulan_num']:
    # kalau bulannya mundur (misalnya dari 12 ke 1), berarti sudah ganti tahun
    if bulan < bulan_sebelumnya:
        tahun_sekarang = tahun_sekarang + 1
    daftar_tahun.append(tahun_sekarang)
    bulan_sebelumnya = bulan

cek['tahun'] = daftar_tahun

print('Jumlah nasabah per tahun:')
print(cek['tahun'].value_counts().sort_index())
```

### **4.8 Korelasi Antar Variabel Numerik**

Analisis korelasi digunakan untuk melihat seberapa kuat hubungan antara dua variabel numerik. Nilai korelasi berkisar dari −1 hingga 1, dengan tanda positif menunjukkan kecenderungan bergerak searah dan tanda negatif menunjukkan kecenderungan bergerak berlawanan arah.

Variabel target y diubah menjadi angka, yaitu 1 untuk yes dan 0 untuk no, agar hubungan awalnya dengan variabel numerik dapat diperiksa.

```python
# Memeriksa hubungan antarvariabel numerik dan hasil deposito menggunakan korelasi Spearman.
cek_korelasi = df_raw[kolom_num].copy()
cek_korelasi['y'] = df_raw['y'].map({'no': 0, 'yes': 1})      # yes = 1, no = 0
korelasi = cek_korelasi.corr(method='spearman')

plt.figure(figsize=(8, 6))
sns.heatmap(korelasi, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Korelasi Spearman antar kolom numerik dan y')
plt.tight_layout()
plt.show()
```

**Hasil Analisis Korelasi**

- **`pdays` dan `previous` memiliki korelasi sangat kuat (0,99).** Keduanya sama-sama menggambarkan riwayat kontak nasabah. Nasabah yang belum pernah dihubungi memiliki nilai `pdays` = -1 dan `previous` = 0. Karena informasinya saling berkaitan, analisis cukup menggunakan satu indikator riwayat kontak.

- **`duration` memiliki hubungan paling kuat dengan y (0,34).** Semakin lama percakapan, semakin tinggi kecenderungan nasabah membuka deposito. Namun, durasi tidak digunakan untuk menentukan prioritas nasabah karena baru diketahui setelah panggilan selesai.

- **Riwayat kampanye dan saldo memiliki hubungan positif yang lemah dengan y.** Nilai korelasi `previous` sebesar 0,17, `pdays` sebesar 0,15, dan `balance` sebesar 0,10.

- **`campaign` memiliki hubungan negatif yang lemah dengan y (-0,08).** Semakin banyak panggilan yang dilakukan, kecenderungan nasabah membuka deposito terlihat menurun.

- **`age` memiliki korelasi hampir nol (-0,01).** Namun, ini bukan berarti usia tidak berpengaruh. Grafik sebelumnya menunjukkan bahwa peluang membuka deposito berbeda menurut kelompok usia, dengan pola menyerupai huruf U. Korelasi biasa tidak dapat menggambarkan pola seperti ini dengan baik.

- **Tidak ada pasangan kolom numerik lain yang menunjukkan korelasi kuat.** Selain `pdays` dan `previous`, korelasi antarkolom relatif rendah sehingga tidak ada kolom yang perlu dihapus hanya karena informasinya berulang.

### **5.6 Feature Engineering**

Feature engineering adalah proses membuat kolom baru dari data yang sudah ada agar pola dan hubungan dalam data lebih mudah dianalisis.

```python
# Membuat nomor bulan dan kolom tahun
peta_bulan = {'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6,
              'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12}
df['bulan_num'] = df['month'].astype(str).map(peta_bulan)

# tahun bertambah satu setiap kali bulannya mundur (misalnya dari dec ke jan)
daftar_tahun = []
tahun_sekarang = 2008
bulan_sebelumnya = df['bulan_num'].iloc[0]
for bulan in df['bulan_num']:
    if bulan < bulan_sebelumnya:
        tahun_sekarang = tahun_sekarang + 1
    daftar_tahun.append(tahun_sekarang)
    bulan_sebelumnya = bulan
df['tahun'] = daftar_tahun

hasil = df['tahun'].value_counts().sort_index().to_frame('jumlah_nasabah')
hasil['persen_%'] = (hasil['jumlah_nasabah'] / len(df) * 100).round(1)
hasil
```

Kolom bulan_num dan tahun ditambahkan untuk memudahkan analisis berdasarkan waktu. Karena data awal tidak mencantumkan tahun, bulan yang sama dari tahun berbeda bisa tercampur, padahal hasil kampanyenya berbeda.

```python
# Mengelompokkan usia nasabah agar pola keberhasilan deposito lebih mudah dibandingkan.
df['kelompok_usia'] = pd.cut(df['age'], bins=[17, 24, 34, 44, 54, 64, 120],
                             labels=['18-24', '25-34', '35-44', '45-54', '55-64', '65+'])

hasil = df['kelompok_usia'].value_counts(sort=False).to_frame('jumlah_nasabah')
hasil['persen_%'] = (hasil['jumlah_nasabah'] / len(df) * 100).round(1)
hasil
```

Usia dikelompokkan dalam rentang 10 tahun karena tingkat keberhasilan deposito cenderung tinggi pada usia muda dan tua, tetapi lebih rendah di usia tengah (§4.4). Pola ini lebih mudah terlihat jika usia dikelompokkan.

```python
# Menggabungkan informasi KPR dan pinjaman pribadi menjadi kategori beban pinjaman.
df['beban_pinjaman'] = 'KPR + pinjaman pribadi'
df.loc[(df['housing'] == 'no') & (df['loan'] == 'no'), 'beban_pinjaman'] = 'Tanpa pinjaman'
df.loc[(df['housing'] == 'yes') & (df['loan'] == 'no'), 'beban_pinjaman'] = 'KPR saja'
df.loc[(df['housing'] == 'no') & (df['loan'] == 'yes'), 'beban_pinjaman'] = 'Pinjaman pribadi saja'

hasil = df['beban_pinjaman'].value_counts().to_frame('jumlah_nasabah')
hasil['persen_%'] = (hasil['jumlah_nasabah'] / len(df) * 100).round(1)
hasil
```

Kolom housing (KPR) dan loan (pinjaman pribadi) digabung menjadi empat kelompok berdasarkan kepemilikan pinjaman. Tujuannya untuk melihat apakah jenis pinjaman memengaruhi keberhasilan deposito, atau cukup dibedakan berdasarkan ada atau tidaknya pinjaman.

```python
# Mengelompokkan jumlah panggilan dalam kampanye menjadi beberapa tingkat frekuensi kontak.
df['kelompok_kontak'] = pd.cut(df['campaign'], bins=[0, 1, 3, 5, 10, 100],
                               labels=['1', '2-3', '4-5', '6-10', '> 10'])

hasil = df['kelompok_kontak'].value_counts(sort=False).to_frame('jumlah_nasabah')
hasil['persen_%'] = (hasil['jumlah_nasabah'] / len(df) * 100).round(1)
hasil
```

Jumlah panggilan juga dikelompokkan karena tingkat keberhasilan deposito cenderung turun ketika nasabah semakin sering ditelepon (§4.4). Panggilan pertama dipisahkan karena tingkat keberhasilannya paling tinggi, sedangkan panggilan kedua dan ketiga digabung karena hasilnya sama, yaitu 11,2%. Terdapat 1.195 nasabah yang ditelepon lebih dari 10 kali.

```python
# Menentukan periode tahun-bulan dan membedakan kampanye massal dari kampanye selektif berdasarkan jumlah nasabah yang dihubungi.
blok = df['tahun'] * 100 + df['bulan_num']          # penanda tahun-bulan, misalnya 200805
jumlah_bulanan = blok.value_counts()                 # jumlah nasabah tiap bulan

df['mode_kampanye'] = 'Selektif'
df.loc[blok.map(jumlah_bulanan) >= 1000, 'mode_kampanye'] = 'Massal'

pd.DataFrame({
    'keterangan': ['Jumlah bulan massal', 'Jumlah bulan selektif',
                   'Bulan massal terkecil (nasabah)', 'Bulan selektif terbesar (nasabah)',
                   'Nasabah kampanye massal', 'Nasabah kampanye selektif'],
    'nilai': [(jumlah_bulanan >= 1000).sum(), (jumlah_bulanan < 1000).sum(),
              jumlah_bulanan[jumlah_bulanan >= 1000].min(), jumlah_bulanan[jumlah_bulanan < 1000].max(),
              (df['mode_kampanye'] == 'Massal').sum(), (df['mode_kampanye'] == 'Selektif').sum()]
})
```

Jumlah nasabah yang dihubungi berbeda-beda setiap bulan. Untuk membedakannya, bulan dengan 1.000 nasabah atau lebih masuk kategori Massal (9 bulan), sedangkan sisanya masuk kategori Selektif (21 bulan).

Batas 1.000 dipilih karena jumlah nasabah pada kedua kelompok memiliki jarak yang jelas: bulan Selektif terbanyak memiliki 772 nasabah, sedangkan bulan Massal paling sedikit memiliki 1.176 nasabah. Tidak ada bulan dengan jumlah nasabah di antara kedua angka tersebut.

```python
# Menyusun tingkat prioritas nasabah berdasarkan riwayat pembelian dan kondisi keuangan.
tanpa_pinjaman = (df['housing'] == 'no') & (df['loan'] == 'no')
profil_kuat = df['job'].isin(['student', 'retired']) | (df['balance'] >= 1000)

df['prioritas'] = 'P4 - Punya KPR / pinjaman pribadi'
df.loc[tanpa_pinjaman, 'prioritas'] = 'P3 - Tanpa pinjaman, lainnya'
df.loc[tanpa_pinjaman & profil_kuat, 'prioritas'] = 'P2 - Tanpa pinjaman, profil kuat'
df.loc[df['poutcome'] == 'success', 'prioritas'] = 'P1 - Pernah membeli'

jejak.append(('Feature engineering (hanya menambah kolom)', len(df)))

hasil = df['prioritas'].value_counts().sort_index().to_frame('jumlah_nasabah')
hasil['persen_%'] = (hasil['jumlah_nasabah'] / len(df) * 100).round(1)
hasil
```

Prioritas P1–P4 dibuat untuk mengurutkan nasabah berdasarkan peluang membuka deposito, dari yang tertinggi hingga terendah. Dengan begitu, agen bisa menentukan nasabah mana yang perlu dihubungi lebih dulu. Urutan ini didasarkan pada pola yang ditemukan dalam data.

**P1 — Pernah membeli deposito sebelumnya**

**P2 — Memiliki profil yang lebih menjanjikan**

- Saldo minimal €1.000: nasabah dengan saldo pada tingkat ini memiliki tingkat keberhasilan di atas rata-rata keseluruhan, yaitu 11,7%.
- Nasabah yang memenuhi salah satu kriteria tersebut masuk P2, sedangkan sisanya masuk P3.

**P3 — Tidak memiliki pinjaman, tetapi tidak memenuhi kriteria P2**

Kelompok ini tetap memiliki peluang yang lebih baik dibandingkan nasabah yang memiliki pinjaman, tetapi tidak menunjukkan petunjuk tambahan sekuat P2.

**P4 — Memiliki KPR atau pinjaman pribadi**

P4 mencakup kelompok terbesar, yaitu 60,8% dari seluruh nasabah, tetapi tingkat keberhasilannya paling rendah, sebesar 6,8%. Mereka tetap dihubungi, tetapi mendapat prioritas terakhir agar agen dapat lebih dulu berfokus pada nasabah dengan peluang keberhasilan lebih tinggi.

### **6.1 Uji Statistik yang Digunakan**

Uji statistik digunakan untuk mendukung hasil eksplorasi data dengan melihat apakah pola yang ditemukan memiliki hubungan yang cukup kuat secara statistik.

Metode yang digunakan disesuaikan dengan jenis data:

| Jenis Data | Metode | Tujuan |
|---|---|---|
| Kategorikal × Kategorikal | Chi-Square + Cramér's V | Melihat hubungan dan kekuatan hubungan antar variabel kategori |
| Dua Kelompok Proporsi | Uji Dua Proporsi | Membandingkan perbedaan conversion rate antar kelompok |
| Numerik × Dua Kelompok | Mann–Whitney | Membandingkan nilai numerik (saldo dan usia) antara nasabah yang membuka dan yang menolak deposito, karena datanya tidak berdistribusi normal |
| Numerik × Numerik | Korelasi Spearman | Melihat arah dan kekuatan hubungan berdasarkan peringkat, tanpa mensyaratkan distribusi normal |

Tingkat signifikansi yang digunakan adalah α = 0,05. Hasil uji tidak hanya dinilai dari p-value, tetapi juga dari besarnya perbedaan atau kekuatan hubungan. Hal ini penting karena dataset yang besar dapat membuat perbedaan kecil sekalipun terlihat signifikan secara statistik.

Untuk membantu menilai kekuatan hubungan kategorikal, Cramér's V digunakan bersama hasil Chi-square. Sebagai panduan umum, nilai sekitar 0,1 sering dianggap lemah, 0,3 sedang, dan 0,5 kuat, meskipun interpretasi tetap perlu mempertimbangkan konteks analisis (Cohen, 1988).

```python
# Menyiapkan tingkat signifikansi, tempat penyimpanan hasil uji statistik, dan fungsi untuk menghitung KPI.
alpha = 0.05          
catatan_uji = []      

total_nasabah = len(df)
total_deposito = df['y_num'].sum()
total_panggilan = df['campaign'].sum()


def tabel_kpi(data, kolom):
    t = pd.DataFrame({
        'nasabah': data.groupby(kolom)['y_num'].count(),
        'deposito': data.groupby(kolom)['y_num'].sum(),
        'panggilan': data.groupby(kolom)['campaign'].sum()})
    t['CR_%'] = t['deposito'] / t['nasabah'] * 100
    t['panggilan_per_deposito'] = t['panggilan'] / t['deposito']
    t['persentase_nasabah_%'] = t['nasabah'] / t['nasabah'].sum() * 100
    t['persentase_panggilan_%'] = t['panggilan'] / t['panggilan'].sum() * 100
    t['persentase_deposito_%'] = t['deposito'] / t['deposito'].sum() * 100
    return t.round(1)


def tulis_p(p):
    if p < 0.001:
        return '< 0,001'
    return str(round(p, 3))


def uji_chi_square(data, kolom, nama):
    tabel = pd.crosstab(data[kolom], data['y'])
    chi2, p, dof, harapan = stats.chi2_contingency(tabel)
    n = tabel.values.sum()
    v = np.sqrt(chi2 / (n * (min(tabel.shape) - 1)))          # Cramér's V
    catatan_uji.append({'uji': 'Chi-square', 'yang_diuji': nama, 'p_value': p,
                        'ukuran_efek': "Cramér's V = " + str(round(v, 3))})
    return {'yang_diuji': nama, 'chi2': round(chi2, 1), 'p_value': tulis_p(p),
            'cramers_v': round(v, 3), 'frekuensi_harapan_terkecil': round(harapan.min(), 1)}


def uji_dua_proporsi(yes1, n1, yes2, n2, nama):
    p1 = yes1 / n1                                              # CR kelompok 1
    p2 = yes2 / n2                                              # CR kelompok 2
    p_gabungan = (yes1 + yes2) / (n1 + n2)
    z = (p1 - p2) / np.sqrt(p_gabungan * (1 - p_gabungan) * (1 / n1 + 1 / n2))
    p = 2 * stats.norm.sf(abs(z))
    se = np.sqrt(p1 * (1 - p1) / n1 + p2 * (1 - p2) / n2)      # untuk selang kepercayaan 95%
    selisih = (p1 - p2) * 100
    catatan_uji.append({'uji': 'Uji dua proporsi', 'yang_diuji': nama, 'p_value': p,
                        'ukuran_efek': 'selisih = ' + str(round(selisih, 1)) + ' poin'})
    return {'yang_diuji': nama, 'CR_1_%': round(p1 * 100, 1), 'CR_2_%': round(p2 * 100, 1),
            'selisih_poin': round(selisih, 1),
            'batas_bawah_95%': round(selisih - 1.96 * se * 100, 1),
            'batas_atas_95%': round(selisih + 1.96 * se * 100, 1),
            'p_value': tulis_p(p)}
```

## **7.4 BQ4 — Karakteristik Nasabah Seperti Apa yang Paling Berkaitan dengan Pembukaan Deposito**

Sementara itu, variabel default memiliki hubungan paling lemah dengan nilai 0,022. Hal ini menunjukkan bahwa hasil kampanye sebelumnya merupakan variabel yang paling terkait dengan keberhasilan telemarketing di antara variabel yang diuji.

```python
# Menguji hubungan berbagai karakteristik nasabah dengan hasil deposito dan mengurutkan kekuatan hubungannya.
urut = kekuatan.sort_values('cramers_v')
plt.figure(figsize=(8, 4.5))
batang = plt.barh(urut['yang_diuji'], urut['cramers_v'], color='steelblue')
plt.bar_label(batang, fmt='%.3f')
plt.title('Kekuatan hubungan tiap ciri dengan y')
plt.xlabel("Cramér's V (0 = tidak berhubungan, 1 = sempurna)")
plt.tight_layout()
plt.show()
```

```python
# Membandingkan saldo dan usia nasabah yang membuka deposito dengan yang tidak menggunakan uji Mann–Whitney.
baris = []
for kolom in ['balance', 'age']:
    nilai_yes = df[df['y'] == 'yes'][kolom]
    nilai_no = df[df['y'] == 'no'][kolom]
    u, p = stats.mannwhitneyu(nilai_yes, nilai_no)
    catatan_uji.append({'uji': 'Mann-Whitney', 'yang_diuji': kolom + ': nasabah yes vs no', 'p_value': p,
                        'ukuran_efek': 'median ' + str(nilai_yes.median()) + ' vs ' + str(nilai_no.median())})
    baris.append({'kolom': kolom, 'median_yes': nilai_yes.median(), 'median_no': nilai_no.median(), 'p_value': tulis_p(p)})
pd.DataFrame(baris)
```

```python
# Menghitung kontribusi kumulatif nasabah, panggilan, dan deposito saat daftar diprioritaskan dari P1 hingga P4.
for nama, data in [('Semua data', df), ('Kampanye massal', massal)]:
    t = tabel_kpi(data, 'prioritas')[['nasabah', 'deposito', 'panggilan']]
    # cumsum = jumlah berjalan: P1, lalu P1+P2, lalu P1+P2+P3, dan seterusnya
    t['nasabah_kumulatif_%'] = (t['nasabah'].cumsum() / t['nasabah'].sum() * 100).round(1)
    t['deposito_kumulatif_%'] = (t['deposito'].cumsum() / t['deposito'].sum() * 100).round(1)
    t['CR_kumulatif_%'] = (t['deposito'].cumsum() / t['nasabah'].cumsum() * 100).round(1)
    t['panggilan_per_deposito_kumulatif'] = (t['panggilan'].cumsum() / t['deposito'].cumsum()).round(1)
    print(nama)
    display(t)
```

Hasil uji chi-square menunjukkan bahwa tingkat prioritas berhubungan nyata dengan keputusan nasabah membuka deposito, baik pada seluruh data maupun pada masing-masing cara kampanye (semua p-value < 0,001). Kekuatan hubungannya sedang pada seluruh data (Cramér’s V = 0,339), mendekati sedang pada kampanye selektif (0,277), dan lemah pada kampanye massal (0,118). Meskipun begitu, urutan conversion rate P1 sampai P4 tetap konsisten, sehingga pembagian prioritas layak dipakai untuk menentukan nasabah yang ditelepon lebih dulu.

## **7.6 BQ6 — Berapa panggilan yang bisa dihemat dengan aturan prioritas dan batas telepon?**

Aturan baru diuji menggunakan data historis. Jika batas telepon ditetapkan sebanyak k kali, panggilan kepada nasabah yang sebelumnya dihubungi lebih dari k kali dihitung hanya sampai batas tersebut.

Simulasi menggunakan asumsi yang hati-hati: deposito yang sebelumnya terjadi setelah telepon ke-k dianggap hilang seluruhnya. Karena itu, estimasi kehilangan deposito merupakan batas atas, bukan kepastian bahwa seluruh deposito tersebut benar-benar akan hilang. Batas telepon dipilih dengan syarat setidaknya 85% deposito tetap dipertahankan.

```python
# Membuat perhitungan simulasi untuk melihat penghematan panggilan dan jumlah deposito yang dipertahankan pada berbagai batas telepon.
def tabel_batas(data):
    baris = []
    for batas in [1, 2, 3, 4, 5, 6]:
        panggilan = data['campaign'].clip(upper=batas).sum()
        deposito = ((data['campaign'] <= batas) & (data['y_num'] == 1)).sum()
        dipotong = data['campaign'].sum() - panggilan
        hilang = data['y_num'].sum() - deposito
        baris.append({'batas_telepon': batas, 'panggilan': panggilan, 'deposito': deposito,
                      'panggilan_dihemat_%': round(dipotong / data['campaign'].sum() * 100, 1),
                      'deposito_dipertahankan_%': round(deposito / data['y_num'].sum() * 100, 1),
                      'panggilan_per_deposito': round(panggilan / deposito, 1),
                      'hasil_per_100_panggilan_dipotong': round(hilang / dipotong * 100, 2)})
    return pd.DataFrame(baris).set_index('batas_telepon')

batas_semua = tabel_batas(df)
display(batas_semua)
```

Tabel ini mensimulasikan apa yang terjadi bila setiap nasabah hanya boleh ditelepon sampai batas tertentu. Semakin ketat batasnya, semakin banyak panggilan yang dihemat, tetapi semakin banyak pula deposito yang hilang. Batas 2 telepon menghemat 41,7% panggilan, tetapi hanya mempertahankan 74,9% deposito, sedangkan batas 3 telepon menghemat 29,5% panggilan dan tetap mempertahankan 86,6% deposito. Jadi, batas 3 telepon adalah pilihan paling hemat yang masih memenuhi syarat minimal 85% deposito dipertahankan.

```python
# Memvisualisasikan pertukaran antara panggilan yang dihemat dan deposito yang dipertahankan di tiap batas telepon
plt.figure(figsize=(8, 4))
plt.plot(batas_semua.index, batas_semua['deposito_dipertahankan_%'], marker='o', color='steelblue', label='Deposito dipertahankan (%)')
plt.plot(batas_semua.index, batas_semua['panggilan_dihemat_%'], marker='o', color='darkorange', label='Panggilan dihemat (%)')
plt.axhline(85, color='gray', linestyle='--')
plt.title('Memilih batas telepon (garis = 85%)')
plt.xlabel('Batas jumlah telepon per nasabah')
plt.ylabel('Persen')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Mensimulasikan beberapa aturan batas panggilan dan membandingkan penghematan serta risiko kehilangan deposito.
# B: semua nasabah paling banyak 3 telepon; deposito setelah telepon ke-3 dianggap hilang
df['panggilan_B'] = df['campaign'].clip(upper=3)
df['deposito_B'] = 0
df.loc[df['campaign'] <= 3, 'deposito_B'] = df['y_num']

# C: seperti B, tetapi P1 tidak dibatasi
df['panggilan_C'] = df['panggilan_B']
df['deposito_C'] = df['deposito_B']
df.loc[df['tingkat'] == 'P1', 'panggilan_C'] = df['campaign']
df.loc[df['tingkat'] == 'P1', 'deposito_C'] = df['y_num']

# D: seperti C, tetapi P4 paling banyak 1 telepon
df['panggilan_D'] = df['panggilan_C']
df['deposito_D'] = df['deposito_C']
df.loc[df['tingkat'] == 'P4', 'panggilan_D'] = 1
df.loc[(df['tingkat'] == 'P4') & (df['campaign'] > 1), 'deposito_D'] = 0

# E: seperti C, tetapi P4 tidak ditelepon
df['panggilan_E'] = df['panggilan_C']
df['deposito_E'] = df['deposito_C']
df.loc[df['tingkat'] == 'P4', 'panggilan_E'] = 0
df.loc[df['tingkat'] == 'P4', 'deposito_E'] = 0

daftar_skenario = [('A. Kondisi sekarang', 'campaign', 'y_num'),
                   ('B. Semua maks. 3 telepon', 'panggilan_B', 'deposito_B'),
                   ('C. P1 bebas, P2-P4 maks. 3', 'panggilan_C', 'deposito_C'),
                   ('D. Seperti C, P4 maks. 1', 'panggilan_D', 'deposito_D'),
                   ('E. Seperti C, P4 tidak ditelepon', 'panggilan_E', 'deposito_E')]

baris = []
for nama, kolom_panggilan, kolom_deposito in daftar_skenario:
    panggilan = df[kolom_panggilan].sum()
    deposito = df[kolom_deposito].sum()
    nasabah_ditelepon = (df[kolom_panggilan] > 0).sum()
    baris.append({'skenario': nama, 'panggilan': panggilan, 'deposito': deposito,
                  'panggilan_per_deposito': round(panggilan / deposito, 1),
                  'deposito_per_100_panggilan': round(deposito / panggilan * 100, 2),
                  'CR_%': round(deposito / nasabah_ditelepon * 100, 1),
                  'perubahan_panggilan_%': round((panggilan / total_panggilan - 1) * 100, 1),
                  'deposito_dipertahankan_%': round(deposito / total_deposito * 100, 1)})
skenario = pd.DataFrame(baris).set_index('skenario')
display(skenario)
```

## **8. Business Metric: Interpretasi KPI**

Bagian ini merangkum metrik bisnis yang digunakan untuk menilai efektivitas kampanye, baik pada kondisi awal maupun setelah simulasi penerapan aturan baru.

Setiap metrik memberikan sudut pandang yang berbeda. Panggilan per deposito menunjukkan efisiensi penggunaan sumber daya, conversion rate menunjukkan keberhasilan penawaran, sedangkan deposito yang dipertahankan menunjukkan apakah penghematan panggilan berisiko mengurangi hasil kampanye secara berlebihan.

```python
# Menghitung KPI bisnis dengan rumus yang sama untuk berbagai kelompok nasabah dan jenis kampanye.
daftar_kelompok = [
    ('Keseluruhan', df),
    ('Kampanye massal', massal),
    ('Kampanye selektif', selektif),
    ('Ditelepon 1-3 kali', df[df['campaign'] <= 3]),
    ('Ditelepon lebih dari 3 kali', df[df['campaign'] > 3]),
    ('P1 - Pernah membeli', df[df['tingkat'] == 'P1']),
    ('P2 - Tanpa pinjaman, profil kuat', df[df['tingkat'] == 'P2']),
    ('P3 - Tanpa pinjaman, lainnya', df[df['tingkat'] == 'P3']),
    ('P4 - Punya KPR / pinjaman pribadi', df[df['tingkat'] == 'P4']),
]

baris = []
for nama, data in daftar_kelompok:
    nasabah = len(data)
    deposito = data['y_num'].sum()
    panggilan = data['campaign'].sum()
    panggilan_penolak = data[data['y'] == 'no']['campaign'].sum()
    baris.append({'kelompok': nama, 'nasabah': nasabah, 'panggilan': panggilan, 'deposito': deposito,
                  'conversion_rate_%': round(deposito / nasabah * 100, 1),
                  'panggilan_per_deposito': round(panggilan / deposito, 1),
                  'deposito_per_100_panggilan': round(deposito / panggilan * 100, 2),
                  'persentase_panggilan_ke_penolak_%': round(panggilan_penolak / panggilan * 100, 1),
                  'lift': round((deposito / nasabah) / (total_deposito / total_nasabah), 2)})

# skenario C (rekomendasi) sebagai pembanding
nasabah = len(df)
panggilan = df['panggilan_C'].sum()
deposito = df['deposito_C'].sum()
baris.append({'kelompok': 'Skenario C (simulasi)', 'nasabah': nasabah, 'panggilan': panggilan, 'deposito': deposito,
              'conversion_rate_%': round(deposito / nasabah * 100, 1),
              'panggilan_per_deposito': round(panggilan / deposito, 1),
              'deposito_per_100_panggilan': round(deposito / panggilan * 100, 2),
              'persentase_panggilan_ke_penolak_%': np.nan,
              'lift': round((deposito / nasabah) / (total_deposito / total_nasabah), 2)})
pd.DataFrame(baris).set_index('kelompok')
```

## **11. Batasan Project & Future Work**

### **11.1 Batasan**

Meskipun analisis menghasilkan beberapa rekomendasi yang dapat digunakan untuk meningkatkan efisiensi kampanye, terdapat keterbatasan yang perlu dipertimbangkan ketika menginterpretasikan hasilnya.

1. **Dampak berasal dari simulasi historis.** Estimasi penghematan dihitung dengan menerapkan aturan baru pada data lama. Hasil kampanye nyata dapat berbeda karena respons nasabah dan perilaku agen mungkin berubah.
2. **Perhitungan potensi kehilangan deposito menggunakan asumsi konservatif.** Simulasi menganggap deposito yang terjadi setelah batas panggilan sebagai hasil yang akan hilang jika panggilan lanjutan dihentikan. Dalam praktiknya, sebagian nasabah mungkin tetap membuka deposito melalui panggilan yang lebih sedikit. Besarnya perbedaan tersebut belum dapat dipastikan dari data yang tersedia.

3. **Aturan prioritas dibuat dan dievaluasi menggunakan dataset yang sama.** Pemeriksaan berdasarkan tahun dan pembagian data secara acak membantu menilai kestabilan pola, tetapi belum menggantikan pengujian pada kampanye baru yang tidak digunakan dalam pembentukan aturan.

4. **Pengaruh cara kampanye belum dapat dipisahkan sepenuhnya dari pengaruh waktu.** Sebagian besar kampanye pada 2008 tergolong massal, sedangkan kampanye pada 2010 tergolong selektif. Kondisi ekonomi juga berubah selama periode pengamatan. Karena data tidak memuat seluruh faktor ekonomi yang relevan, perbedaan hasil kedua kampanye tidak dapat langsung diartikan sebagai hubungan sebab-akibat.

5. **Data biaya dan nilai deposito tidak tersedia.** Dampak rekomendasi belum dapat diterjemahkan menjadi penghematan biaya dalam euro atau tambahan pendapatan yang sebenarnya. Simulasi juga memperlakukan setiap deposito sebagai satu hasil dengan bobot yang sama.

6. **Tidak tersedia riwayat lengkap setiap panggilan.** Data hanya menyediakan durasi percakapan terakhir, sehingga strategi berdasarkan durasi panggilan sebelumnya tidak dapat diuji secara memadai.

7. **ID nasabah dan kolom tahun tidak tersedia secara langsung.** Tahun direkonstruksi berdasarkan urutan data, sedangkan nasabah tidak dapat dilacak secara langsung dari satu kampanye ke kampanye berikutnya.

8. **Beberapa kelompok memiliki jumlah observasi yang terbatas.** Sebagai contoh, P1 pada kampanye massal hanya mencakup 306 nasabah, dan P1 yang dihubungi lebih dari tiga kali hanya mencakup 123 nasabah. Hasil pada kelompok kecil lebih rentan berubah ketika diterapkan pada data lain.

9. **Data berasal dari satu bank di Portugal pada 2008–2010.** Pola yang ditemukan belum tentu berlaku dengan kekuatan yang sama pada bank, negara, periode, atau kondisi pasar yang berbeda.

### **11.2 Future Work**

Untuk meningkatkan keandalan hasil dan memperkuat penerapan rekomendasi, beberapa pengembangan dapat dilakukan.

1. **Melakukan A/B test pada kampanye nyata.** Aturan prioritas dan batas panggilan diuji dengan membandingkan kelompok yang menggunakan strategi lama dan strategi baru. Langkah ini memungkinkan bank mengukur dampak aktual terhadap efisiensi dan jumlah deposito.

2. **Menguji kampanye selektif secara lebih terkontrol.** Kampanye dengan jumlah nasabah dan karakteristik yang sebanding dapat dijalankan untuk menilai apakah pendekatan selektif benar-benar meningkatkan conversion rate atau apakah perbedaannya turut dipengaruhi kondisi waktu dan faktor lainnya.

3. **Menambahkan informasi biaya panggilan dan nilai deposito.** Dengan informasi ini, bank dapat menghitung penghematan finansial, memperkirakan manfaat dari tambahan deposito, dan menentukan batas panggilan berdasarkan pertimbangan biaya dan hasil bisnis.

4. **Melengkapi data ekonomi dan informasi sumber daftar kampanye.** Informasi seperti suku bunga, tanggal lengkap, ID nasabah, dan asal daftar dapat membantu memisahkan pengaruh kondisi ekonomi dari pengaruh pemilihan nasabah.

5. **Menguji temuan mengenai hari Senin.** Conversion rate yang lebih rendah pada hari Senin dalam kampanye selektif dapat diteliti lebih lanjut. Namun, pola ini perlu dikonfirmasi pada kampanye tambahan sebelum digunakan untuk menentukan jadwal operasional.

6. **Mengembangkan model prediksi sederhana.** Regresi logistik dapat dipertimbangkan untuk memperkirakan peluang seorang nasabah membuka deposito berdasarkan informasi yang tersedia sebelum panggilan. Model ini dapat dibandingkan dengan aturan P1–P4 untuk menilai apakah peningkatan ketepatan prediksi sebanding dengan kompleksitas implementasinya.

7.  **Membangun dashboard pemantauan kampanye.** Dashboard dapat menampilkan conversion rate, panggilan per deposito, jumlah panggilan berulang, hasil setiap tingkat prioritas, dan perbandingan kinerja antarperiode. Dengan begitu, pengelola kampanye dapat mengevaluasi hasil secara berkala dan menyesuaikan strategi berdasarkan data terbaru.
